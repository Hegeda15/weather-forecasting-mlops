```python
import os
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv
```

# Loading data with SQL query into a DataFrame

```python
load_dotenv()
engine = create_engine(os.getenv("SUPABASE_DB_URL"))

query = "SELECT * FROM weather_logs ORDER BY timestamp ASC;"
df = pd.read_sql(query, con=engine)
df.head()
```
