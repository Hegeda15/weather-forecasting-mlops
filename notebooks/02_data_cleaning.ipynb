```python
import os
import pandas as pd
from sqlalchemy import create_engine
from dotenv import load_dotenv
import numpy as np
```

```python
load_dotenv()
engine = create_engine(os.getenv("SUPABASE_DB_URL"))

query = "SELECT * FROM weather_logs ORDER BY timestamp ASC;"
df = pd.read_sql(query, con=engine)
df.head()
```

```python
cols=["temperature_2m","surface_pressure","wind_speed_10m","apparent_temperature","wind_gusts_10m","rain","precipitation"]
df[cols]=df[cols].round(2)
df.head()
```

```python
df=df.drop("created_at",axis=1)
df.head()
```

```python
df["timestamp"] = df["timestamp"].dt.strftime("%Y-%m-%d %H:%M")
df.head()
```

```python
table_name = "weather_logs_cleaned"
df.to_sql(
    name=table_name,
    con=engine,
    if_exists="replace",  # ha már létezik a tábla, írja felül (futtathatod többször is)
    index=False,           # ne mentse el a pandas indexet külön oszlopként
    method="multi"         # felgyorsítja a nagy mennyiségű adat beszúrását
)
```
